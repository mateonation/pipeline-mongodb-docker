Primero que todo, generamos las conexiones necesarias a Mongos y a Mysql para poder trabajar.

```python
from pymongo import MongoClient
uri = "mongodb://root:changeme@localhost:27017/admin"
client = MongoClient(uri)
db = client.pipe

```

```python
import mysql.connector

cnx = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="changeme",
    database="pipe")
```

Establecemos un cursor para poder acceder a la tabla de track. Posteriomente creamos una colección donde isertaremos los documentos.

```python
cursork = cnx.cursor()   
query = "SELECT * FROM track;"
cursork.execute(query)
rows = cursork.fetchall()
```

```python
coleccion = db["tracks-MongoDB"]            
```

```python
for r in rows:
    documento = {
        "_id": r[0],
        "track_name": r[1],
        "artists": r[2],
        "album_name ": r[3],
        "popularity": r[4],
        "duration_ms": r[5]
    }
    coleccion.insert_one(documento)

print("Datos insertados")
```

Realizaremos consultas donde buscamos que se nos muestren todos los documentos insertados(coleccion.find()).También que podamos ver un único documento(coleccion.find_one) y por último un recuento preciso de documentos que estan en la coleccion(coleccion.count_documents({})).

```python
for doc in coleccion.find():
    print(doc)
```

```python
print(coleccion.find_one({"track_name": "Hunger"}))
```

```python
print(coleccion.count_documents({}))
```

```python
consulta1 = coleccion.find_one({"track_name": "Hunger"})
print(consulta1)
```
