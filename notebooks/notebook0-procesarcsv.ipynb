```python
import pandas as pd
```

```python
df = pd.read_csv('../spotify-tracks.csv')
df.sample(5)
```

## EXPLORANDO LOS DATOS

    Con esta función estaremos calculando que valores son nulos de cada columna.

```python
df.columns
```

```python
df.drop(columns="Unnamed: 0", inplace=True)
```

```python
df.isna().sum()
```

Después de utilizar la funcion `isna` vemos que existen 3 columnas con valor nulo, presuponemos que son la misma fila. Extraemos la fila: 

```python
df[df['album_name'].isna()]
```

Tras confirmar que los tres valores nulos pertenecen a la misma fila, decidimos limpiarlos. A continuacion tratamos de ver si existen datos duplicados.

```python
df.dropna(inplace=True)
```

```python
df[df['album_name'].isna()]
```

```python
df = df.iloc[:50000, :6]
```

En el tratamiento inicial de los datos no detectamos duplicados, sin embargo tras trabajar con eld ataset nos dimos cuenta de los duplicados en el track_id.

```python
df.duplicated().sum()
```

```python
df.duplicated(subset="track_id").sum()
```

```python
df.drop_duplicates(subset="track_id",inplace=True)
```

```python
df.to_csv("../spotify-clean.csv", index=False)
```
